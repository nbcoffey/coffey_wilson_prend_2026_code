```python
# May 9 2026 update: making the magnitude the same by rescaling
import numpy as np
import matplotlib.pyplot as plt
from scipy.interpolate import make_interp_spline

# Basic Gaussian perturbation function with sponge layer protection
def create_gaussian_perturbation(yy, mu, sigma, amplitude, bounds=None, sponge_start=None):
    """
    Create a Gaussian perturbation with optional bounds and sponge layer protection
    
    Parameters:
    - yy: y-coordinates (in meters)
    - mu: center of Gaussian (in km)
    - sigma: width of Gaussian (in km)
    - amplitude: maximum strength of perturbation (N/m²)
    - bounds: optional (y_min, y_max) tuple in km to truncate the Gaussian
    - sponge_start: y-coordinate where sponge layer begins (in meters)
    
    Returns:
    - perturbation: wind stress perturbation array
    """
    # Convert to km for calculation
    y_km = yy/1e3
    
    # Create Gaussian
    perturbation = amplitude * np.exp(-np.power((y_km - mu) / sigma, 2.0) / 2)
    
    # Apply bounds if specified
    if bounds is not None:
        y_min, y_max = bounds
        perturbation[(y_km < y_min) | (y_km > y_max)] = 0
    
    # Protect the sponge layer
    if sponge_start is not None:
        perturbation[yy >= sponge_start] = 0
        
    return perturbation

def create_contraction_with_preserved_max(yy, base_profile, shift_distance=700,
                                          new_center_offset=-100, sigma=250,
                                          sponge_start=None, tolerance=1e-8):
    """
    Iteratively finds the correct amplitude to preserve maximum westerly value
    """
    from scipy.optimize import fsolve
    
    westerly_mask = base_profile >= 0
    original_max = np.max(base_profile[westerly_mask])
    max_idx = np.where(westerly_mask)[0][np.argmax(base_profile[westerly_mask])]
    max_y = yy[max_idx]/1e3
    new_center = max_y + new_center_offset
    
    def objective(amplitude):
        """Find amplitude that gives us the original maximum"""
        south = create_gaussian_perturbation(yy, new_center - shift_distance/2,
                                            sigma, amplitude, sponge_start=sponge_start)
        north = create_gaussian_perturbation(yy, new_center + shift_distance/2,
                                            sigma, -amplitude, sponge_start=sponge_start)
        combined = south + north
        new_profile = base_profile + combined
        new_max = np.max(new_profile[westerly_mask])
        return new_max - original_max
    
    # Solve for the amplitude
    optimal_amp = fsolve(objective, 0.015)[0]
    
    # Create final perturbations
    south_pert = create_gaussian_perturbation(yy, new_center - shift_distance/2,
                                              sigma, optimal_amp, sponge_start=sponge_start)
    north_pert = create_gaussian_perturbation(yy, new_center + shift_distance/2,
                                              sigma, -optimal_amp, sponge_start=sponge_start)
    combined_pert = south_pert + north_pert
    
    # Verify
    final_profile = base_profile + combined_pert
    final_max = np.max(final_profile[westerly_mask])
    print(f"Original max: {original_max:.6f}, Final max: {final_max:.6f}")
    print(f"Optimal amplitude: {optimal_amp:.6f}")
    
    return south_pert, north_pert, combined_pert

# Base profile
base_profile = taux.copy()

# Define sponge layer start position
sponge_start = Ly - Ln

# # Create the components of the contraction experiment
# south_pert, north_pert, combined_pert = create_contraction_components(
#     yy, base_profile, shift_distance=700, new_center_offset=-100, sigma=250, sponge_start=sponge_start
# )
# Create the components focusing on westerly winds only
south_pert, north_pert, combined_pert = create_contraction_with_preserved_max(
    yy, base_profile, 
    shift_distance=700, 
    new_center_offset=-100, 
    sigma=250, 
    sponge_start=sponge_start
)

# Visualization of wind stress profiles for contraction components
fig, axes = plt.subplots(1, 3, figsize=(18, 6))
fz = 12  # Font size

# 1. Southern Gaussian (positive perturbation)
ax = axes[0]
ax.plot(base_profile, yy/1e3, 'b-', label='Control', linewidth=2)
ax.plot(south_pert, yy/1e3, 'r--', label='South Gaussian', linewidth=2)
ax.plot(base_profile + south_pert, yy/1e3, 'g-', label='Control + South Gaussian', linewidth=2)
ax.set_title('Southern Component of Contraction', fontsize=fz+2)
ax.set_xlabel('Wind stress (N/m²)', fontsize=fz)
ax.set_ylabel('Y (km)', fontsize=fz)
ax.axvline(0, linestyle='--', linewidth=1, color='k')
ax.axhline(Ys/1e3, linestyle='--', linewidth=1, color='0.5', label='Shelf break')
ax.axhline(sponge_start/1e3, linestyle='-.', linewidth=1, color='r', label='Sponge layer')
ax.set_xlim(-0.2, 0.2)
ax.set_ylim(0, Ly/1e3)
ax.grid(True)
ax.legend(fontsize=fz-2)

# 2. Northern Gaussian (negative perturbation)
ax = axes[1]
ax.plot(base_profile, yy/1e3, 'b-', label='Control', linewidth=2)
ax.plot(north_pert, yy/1e3, 'r--', label='North Gaussian', linewidth=2)
ax.plot(base_profile + north_pert, yy/1e3, 'g-', label='Control + North Gaussian', linewidth=2)
ax.set_title('Northern Component of Contraction', fontsize=fz+2)
ax.set_xlabel('Wind stress (N/m²)', fontsize=fz)
ax.set_ylabel('Y (km)', fontsize=fz)
ax.axvline(0, linestyle='--', linewidth=1, color='k')
ax.axhline(Ys/1e3, linestyle='--', linewidth=1, color='0.5', label='Shelf break')
ax.axhline(sponge_start/1e3, linestyle='-.', linewidth=1, color='r', label='Sponge layer')
ax.set_xlim(-0.2, 0.2)
ax.set_ylim(0, Ly/1e3)
ax.grid(True)
ax.legend(fontsize=fz-2)

# 3. Combined Contraction (both Gaussians)
ax = axes[2]
ax.plot(base_profile, yy/1e3, 'b-', label='Control', linewidth=2)
ax.plot(combined_pert, yy/1e3, 'r--', label='Combined Perturbation', linewidth=2)
ax.plot(base_profile + combined_pert, yy/1e3, 'g-', label='Control + Perturbation', linewidth=2)
ax.set_title('Complete Contraction (Both Gaussians)', fontsize=fz+2)
ax.set_xlabel('Wind stress (N/m²)', fontsize=fz)
ax.set_ylabel('Y (km)', fontsize=fz)
ax.axvline(0, linestyle='--', linewidth=1, color='k')
ax.axhline(Ys/1e3, linestyle='--', linewidth=1, color='0.5', label='Shelf break')
ax.axhline(sponge_start/1e3, linestyle='-.', linewidth=1, color='r', label='Sponge layer')
ax.set_xlim(-0.2, 0.2)
ax.set_ylim(0, Ly/1e3)
ax.grid(True)
ax.legend(fontsize=fz-2)

plt.tight_layout()
# plt.savefig('contraction_components_wind_stress.png', dpi=300, bbox_inches='tight')
plt.show()

# Visualization of wind stress curl for contraction components
fig, axes = plt.subplots(1, 3, figsize=(18, 6))

# Calculate wind stress curl for all profiles
curl_base = np.gradient(base_profile, yy, edge_order=2)
curl_south = np.gradient(south_pert, yy, edge_order=2)
curl_south_total = np.gradient(base_profile + south_pert, yy, edge_order=2)
curl_north = np.gradient(north_pert, yy, edge_order=2)
curl_north_total = np.gradient(base_profile + north_pert, yy, edge_order=2)
curl_combined = np.gradient(combined_pert, yy, edge_order=2)
curl_combined_total = np.gradient(base_profile + combined_pert, yy, edge_order=2)

# Scale factor for better visualization
scale = 1e7

# 1. Southern Gaussian - Curl
ax = axes[0]
ax.plot(-curl_base*scale, yy/1e3, 'b-', label='Control', linewidth=2)
ax.plot(-curl_south*scale, yy/1e3, 'r--', label='South Gaussian', linewidth=2)
ax.plot(-curl_south_total*scale, yy/1e3, 'g-', label='Control + South Gaussian', linewidth=2)
ax.set_title('Wind Stress Curl: Southern Component', fontsize=fz+2)
ax.set_xlabel('Wind stress curl (1E-7 N/m³)', fontsize=fz)
ax.set_ylabel('Y (km)', fontsize=fz)
ax.axvline(0, linestyle='--', linewidth=1, color='k')
ax.axhline(Ys/1e3, linestyle='--', linewidth=1, color='0.5', label='Shelf break')
ax.axhline(sponge_start/1e3, linestyle='-.', linewidth=1, color='r', label='Sponge layer')
ax.set_xlim(-6, 6)
ax.set_ylim(0, Ly/1e3)
ax.grid(True)
ax.legend(fontsize=fz-2)

# 2. Northern Gaussian - Curl
ax = axes[1]
ax.plot(-curl_base*scale, yy/1e3, 'b-', label='Control', linewidth=2)
ax.plot(-curl_north*scale, yy/1e3, 'r--', label='North Gaussian', linewidth=2)
ax.plot(-curl_north_total*scale, yy/1e3, 'g-', label='Control + North Gaussian', linewidth=2)
ax.set_title('Wind Stress Curl: Northern Component', fontsize=fz+2)
ax.set_xlabel('Wind stress curl (1E-7 N/m³)', fontsize=fz)
ax.set_ylabel('Y (km)', fontsize=fz)
ax.axvline(0, linestyle='--', linewidth=1, color='k')
ax.axhline(Ys/1e3, linestyle='--', linewidth=1, color='0.5', label='Shelf break')
ax.axhline(sponge_start/1e3, linestyle='-.', linewidth=1, color='r', label='Sponge layer')
ax.set_xlim(-6, 6)
ax.set_ylim(0, Ly/1e3)
ax.grid(True)
ax.legend(fontsize=fz-2)

# 3. Combined Contraction - Curl
ax = axes[2]
ax.plot(-curl_base*scale, yy/1e3, 'b-', label='Control', linewidth=2)
ax.plot(-curl_combined*scale, yy/1e3, 'r--', label='Combined Perturbation', linewidth=2)
ax.plot(-curl_combined_total*scale, yy/1e3, 'g-', label='Control + Perturbation', linewidth=2)
ax.set_title('Wind Stress Curl: Complete Contraction', fontsize=fz+2)
ax.set_xlabel('Wind stress curl (1E-7 N/m³)', fontsize=fz)
ax.set_ylabel('Y (km)', fontsize=fz)
ax.axvline(0, linestyle='--', linewidth=1, color='k')
ax.axhline(Ys/1e3, linestyle='--', linewidth=1, color='0.5', label='Shelf break')
ax.axhline(sponge_start/1e3, linestyle='-.', linewidth=1, color='r', label='Sponge layer')
ax.set_xlim(-6, 6)
ax.set_ylim(0, Ly/1e3)
ax.grid(True)
ax.legend(fontsize=fz-2)

plt.tight_layout()
# plt.savefig('contraction_components_wind_stress_curl.png', dpi=300, bbox_inches='tight')
plt.show()

# Print information about each component
print(f"Maximum perturbation from southern Gaussian: {np.max(south_pert):.4f} N/m²")
print(f"Minimum perturbation from southern Gaussian: {np.min(south_pert):.4f} N/m²")
print(f"Maximum perturbation from northern Gaussian: {np.max(north_pert):.4f} N/m²")
print(f"Minimum perturbation from northern Gaussian: {np.min(north_pert):.4f} N/m²")
print(f"Maximum of combined perturbation: {np.max(combined_pert):.4f} N/m²")
print(f"Minimum of combined perturbation: {np.min(combined_pert):.4f} N/m²")

# Find peak locations of the Gaussians
south_peak_idx = np.argmax(south_pert)
north_peak_idx = np.argmin(north_pert)  # Minimum because it's negative
print(f"\nSouthern Gaussian peak location: {yy[south_peak_idx]/1e3:.1f} km")
print(f"Northern Gaussian peak location: {yy[north_peak_idx]/1e3:.1f} km")
print(f"Distance between peaks: {(yy[north_peak_idx] - yy[south_peak_idx])/1e3:.1f} km")
```
